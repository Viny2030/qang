### Foundational Principle
An $n$-qubit quantum state is represented by its **$qg$ expectation values**, $qg_P = \langle P \rangle$ for each Pauli string $P$. A unitary gate $U$ acts via:
$$qg'_P = qg_{U^\dagger P U}$$
