```python
import importlib, subprocess, sys
subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-U", "qang>=0.14.0", "numpy", "matplotlib", "scipy"], check=False)
importlib.invalidate_caches()

import numpy as np
import qang
from qang.core import Qang
from qang import formulation as F

print(f"Referencia completa cargada: qang v{qang.__version__}")
```

```python
# 1. Verificación de anclajes métricos en Qang
anchors = [
    ("Polo Norte (|0>)", 0.0),
    ("Ecuador / Superposición Máxima (|+>)", np.pi / 2),
    ("Polo Sur (|1>)", np.pi),
    ("Estado canónico compuerta T", np.pi / 4),
]

print(f"{'Estado':<45}{'theta (rad)':>12}{'qg_Z':>12}{'qg_S':>12}")
for name, theta in anchors:
    qz = Qang.from_angles(theta, mode="polar").value
    qs = Qang.from_angles(theta, mode="entropic").value
    print(f"{name:<45}{theta:12.4f}{qz:12.4f}{qs:12.4f}")
```

```python
# 2. Regularización de singularidad de gradientes polares (Cap. 12-14)
def inv_jacobian_tikhonov(theta, eps=0.05):
    s = np.sin(theta)
    return -s / (s**2 + eps**2)

print(f"{'theta':>10}{'Raw 1/sin':>15}{'Tikhonov (eps=0.05)':>22}")
for th in [np.pi/2, np.pi/4, 0.1, 0.01, 0.001]:
    raw_val = abs(-1.0 / np.sin(th)) if np.sin(th) > 1e-6 else float('inf')
    reg_val = abs(inv_jacobian_tikhonov(th))
    print(f"{th:10.4f}{raw_val:15.2f}{reg_val:22.4f}")
```
